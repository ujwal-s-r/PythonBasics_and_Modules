Python lets your custom objects behave like built-in Python objects.

ou can create your own type

Suppose you're working with coordinates.

You could create:

class Point:
    def __init__(self, x, y):
        self.x = x
        self.y = y

Now:

p = Point(3, 4)

p is your custom object/data type.

But initially, Python doesn't know what this should mean:

p1 + p2

or:

len(p)

or:

print(p)

That's where dunder methods come in.

To build custom data structures, tensor wrappers, or RL environment wrappers, you need to know how Python implements the Data Model using Dunder (Double Underscore) Methods.

When you write t1 + t2 or t1[0, 1], Python doesn't execute magic; it calls specific dunder methods under the hood.
Code Syntax           CPython Dunder Method
─────────────────────────────────────────────
len(obj)       ──►    obj.__len__()
obj[index]     ──►    obj.__getitem__(index)
obj[idx] = val ──►    obj.__setitem__(idx, val)
a + b          ──►    a.__add__(b)
a @ b          ──►    a.__matmul__(b)  (Matrix multiplication)
print(obj)     ──►    obj.__str__()
obj (in REPL)  ──►    obj.__repr__()

3. Operator Overloading (__add__, __mul__, __matmul__)
You can define custom math operations for your classes:

__add__(self, other): Implements self + other.

__radd__(self, other): Reflected addition (e.g., scalar + tensor when scalar doesn't know how to add a tensor).

__matmul__(self, other): Implements matrix multiplication @.

Custom 2D Mini-Tensor
Build a class MiniTensor2D from scratch that models a 2D dense matrix and implements native Python indexing and addition.

Requirements:

__init__(self, rows: int, cols: int, data: list = None):

Store rows and cols.

Store data internally as a flat 1D list self._data of size rows * cols (zeros by default, or populated from a flat input list).

__getitem__(self, key):

If key is a tuple (r, c) of two integers, compute the flat index via row-major order: flat_idx = r * self.cols + c.

Return self._data[flat_idx].

__setitem__(self, key, value):

If key is (r, c), update self._data[flat_idx] = value.

__add__(self, other):

If other is a MiniTensor2D of identical shape (rows and cols), return a new MiniTensor2D with element-wise added values.

If shapes mismatch, raise ValueError("Shape mismatch for addition").

__repr__(self):

Return f"MiniTensor2D(shape=({self.rows}, {self.cols}))".

```python
class MiniTensor2D:
    
    def __init__(self, rows : int, cols : int, data : list=None):
        self.rows = rows
        self.cols = cols
        
        if data is None:
            self.data = [0]*(rows*cols)
        else:
            
            self.data = list(data)
    
    def __getitem__(self,key):
        match key:
            case (r,c):
                indx = r * self.cols + c
                return self.data[indx]
            case _ :
                raise(ValueError("wrong key"))
    
    def __setitem__(self, key, value):
        match key:
            case (r,c):
                indx = r * self.cols + c
                self.data[indx] = value
            case _ :
                raise(ValueError("wrong key"))
            
    def __add__(self, other):
        if isinstance(other, MiniTensor2D) and self.rows == other.rows and self.cols == other.cols:
            result = []
            for x,y in zip(self.data, other.data):
                result.append(x+y)
            new_t = MiniTensor2D(self.rows, self.cols, result)
            return new_t
        else:
            raise(TypeError("mismatch tensors"))
    
    def __repr__(self):
        return f"MiniTensor2D(shape = ({self.rows}, {self.cols}))"
```

```python
t = MiniTensor2D(2, 2, [1, 2, 3, 4])

print(t[1, 0])
```

```python
t1 = MiniTensor2D(2, 2, [1, 2, 3, 4])
t2 = MiniTensor2D(2, 2, [10, 20, 30, 40])

print(t1[0, 1])  # Should print 2
print(t1[1, 0])  # Should print 3

t1[0, 1] = 99
print(t1[0, 1])  # Should print 99

t3 = t1 + t2
print(t3[0, 0])  # Should print 11 (1 + 10)
print(t3[0, 1])  # Should print 119 (99 + 20)
```
